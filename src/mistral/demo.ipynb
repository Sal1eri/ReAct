```python
from utils.agent import ReactAgent
```

```python
ReAct_Agent_SYSTEM_PROMPT = """
You are a ReAct agent that answers the user's question.

You have access to tools.

Rules:
1. If you need information, call a tool.
2. If you call a tool, output ONLY the tool call.
3. Do NOT include "Final Answer" when calling a tool.
4. When you have enough information, output the final answer.
5. When giving the final answer, DO NOT call any tools.
6. Never output both a tool call and a final answer in the same response.

Final Answer rules:
- The final answer must contain ONLY the answer itself.
- No explanations.
- No extra words.

Output formats:

Tool call:
{"name": "<tool_name>", "arguments": {...}}

Final answer:
Final Answer: <answer>
"""
Agent = ReactAgent(agent_prompt=ReAct_Agent_SYSTEM_PROMPT)
print(Agent.memory)
```

```python
user_message = "What is the weather in China?"
Agent.append_user_message(user_message)
resp = Agent.generate()
```

```python
Agent.memory
print(resp)
print(type(resp))
```

```python
import json

tool_calls = json.loads(resp)
normalized_calls =Agent.append_assistant_tool_calls(tool_calls)

Agent.memory
print(Agent.memory[-1]['tool_calls'])
print(type(Agent.memory[-1]['tool_calls']))
```

```python
Agent.add_tool_result(normalized_calls)
Agent.memory
```
